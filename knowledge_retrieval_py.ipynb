```python
class Knowledge:

    def __init__(self):
        self.data = {
            "python": "Python is a programming language",
            "sql": "SQL is used for databases",
            "ai": "AI means Artificial Intelligence"
        }

    def search(self, question):
        return self.data.get(question.lower(), "Answer not found")


# Create object
k = Knowledge()

# Ask question
q = input("Enter your question: ")

# Display answer
print(k.search(q))
```
