# Python Program to Check Even or Odd

**Prompt:** Write a Python program to check whether a given number is even or odd using an if-else statement.


```python

number = int(input("Enter a number: "))

if number % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```


#Swap two numbers

**Prompt**: Write a Python program to swap two numbers without using a third variable.

```python

a = 10
b = 20

a, b = b, a

print("After swapping:")
print("a =", a)
print("b =", b)
```

#Find the largest of three numbers
**Prompt:** Write a Python program to find the largest among three numbers using if-elif-else.


```python

a = 25
b = 50
c = 15

if a >= b and a >= c:
    print("Largest:", a)
elif b >= a and b >= c:
    print("Largest:", b)
else:
    print("Largest:", c)
```

#Count vowels in a string
**Prompt**: Write a Python program to count the number of vowels in a given string.

```python

text = "Hello Python"
count = 0

for char in text.lower():
    if char in "aeiou":
        count += 1

print("Number of vowels:", count)
```

#Check whether a number is positive, negative, or zero
**Prompt**: Write a Python program to identify whether a number is positive, negative, or zero.

```python

number = -5

if number > 0:
    print("Positive number")
elif number < 0:
    print("Negative number")
else:
    print("Zero")
```

#Find the second-largest number
**Prompt**: Write a Python program to find the second-largest distinct number in a list.

```python

numbers = [10, 25, 40, 25, 30]

unique_numbers = sorted(set(numbers))

if len(unique_numbers) >= 2:
    print("Second largest:", unique_numbers[-2])
else:
    print("No second-largest distinct number")
```

#Calculate the sum of a list
**Prompt**: Write a Python program to calculate the sum of all numbers in a list using a loop.

```python

numbers = [10, 20, 30, 40]
total = 0

for number in numbers:
    total += number

print("Sum:", total)
```

Write a Python example to demonstrate polymorphism using two classes, Bird and Dog, that have the same method but behave differently.

```python
class Bird:
    def speak(self):
        print("Chirp")

class Dog:
    def speak(self):
        print("Bark")
```

#Demonstrate inheritance in Python
**Prompt**: Write a Python program to demonstrate inheritance using a parent class and a child class.

```python

class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

dog = Dog()
dog.speak()
dog.bark()
```
