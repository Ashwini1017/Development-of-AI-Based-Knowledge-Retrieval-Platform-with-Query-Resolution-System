# SQL Interview Questions and Practice

**Objective:** To practice commonly asked SQL interview questions using simple SQL queries and example employee data.


Create the sample database

```python

import sqlite3

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary INTEGER,
    city TEXT
)
""")

cursor.executemany("""
INSERT INTO employees VALUES (?, ?, ?, ?, ?)
""", [
    (1, 'Rahul', 'IT', 50000, 'Bangalore'),
    (2, 'Priya', 'HR', 40000, 'Chennai'),
    (3, 'Amit', 'IT', 60000, 'Mumbai'),
    (4, 'Sneha', 'Finance', 45000, 'Bangalore'),
    (5, 'Kiran', 'HR', 35000, 'Chennai'),
    (6, 'Neha', 'IT', 70000, 'Hyderabad')
])

conn.commit()

print("Employee table created successfully!")
```

#Write a PySpark program to create a SparkSession in Google Colab ?

**Promp**: Write a PySpark program to initialize a SparkSession in Google Colab and print a success message

```python
from pyspark.sql import SparkSession

# Initialize a SparkSession
spark = SparkSession.builder \
    .appName("PySpark SQL Example") \
    .getOrCreate()

print("SparkSession created successfully!")

```

#Write a query to create a table named Students with columns id, name, age, and city, and display all records from the table ?

Prompt: Write a SQL using pyspark to create a table named Students with columns id, name, age, and city, and display all records from the table.

```python
spark.sql("""CREATE TABLE IF NOT EXISTS Students (
    id INT,
    name STRING,
    age INT,
    city STRING
)
USING PARQUET
""")

spark.sql("SELECT * FROM Students").show()


```

Insert sample records


```python

spark.sql("""
INSERT INTO Students VALUES
(1, 'Rahul', 21, 'Bangalore'),
(2, 'Priya', 22, 'Chennai'),
(3, 'Amit', 20, 'Mumbai'),
(4, 'Sneha', 23, 'Hyderabad')
""")

spark.sql("SELECT * FROM Students").show()
```

#Practice more Spark SQL queries
Find students older than 20:

```python
spark.sql("SELECT * FROM Students WHERE age > 20").show()
```

Display students from Bangalore:

```python
spark.sql("SELECT * FROM Students WHERE city = 'Bangalore'").show()
```

#Display students in ascending order of age
**Prompt**: Write a Spark SQL query to sort students from youngest to oldest.

```python

spark.sql("""
SELECT * FROM Students
ORDER BY age ASC
""").show()
```

#Find students whose age is between 20 and 22
**Prompt**: Write a query to retrieve students aged between 20 and 22, inclusive.

```python

spark.sql("""
SELECT *
FROM Students
WHERE age BETWEEN 20 AND 22
""").show()
```

#Add a new column using a query
**Prompt**: Display student names with a category indicating whether they are 21 or older.

```python

spark.sql("""
SELECT name, age,
       CASE
           WHEN age >= 21 THEN 'Senior'
           ELSE 'Junior'
       END AS category
FROM Students
""").show()
```
